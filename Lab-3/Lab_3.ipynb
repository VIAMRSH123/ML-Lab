```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.read_csv("hiring.csv")
df.columns = df.columns.str.strip()

word_to_num = {
    'zero':0, 'one':1, 'two':2, 'three':3, 'four':4,
    'five':5, 'six':6, 'seven':7, 'eight':8,
    'nine':9, 'ten':10, 'eleven':11, 'twelve':12
}

df['experience'] = df['experience'].replace(word_to_num)
df['experience'] = pd.to_numeric(df['experience'], errors='coerce').fillna(0)
df['test_score(out of 10)'] = df['test_score(out of 10)'].fillna(df['test_score(out of 10)'].mean())
df['interview_score(out of 10)'] = df['interview_score(out of 10)'].fillna(df['interview_score(out of 10)'].mean())

X = df[['experience', 'test_score(out of 10)', 'interview_score(out of 10)']]
y = df['salary($)']

model = LinearRegression()
model.fit(X, y)

salary1 = model.predict(pd.DataFrame([[2, 9, 6]], columns=X.columns))
salary2 = model.predict(pd.DataFrame([[12, 10, 10]], columns=X.columns))

print(round(salary1[0], 2))
print(round(salary2[0], 2))
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

df = pd.read_csv("1000_Companies.csv")
df.columns = df.columns.str.strip()

X = df[['R&D Spend', 'Administration', 'Marketing Spend', 'State']]
y = df['Profit']

ct = ColumnTransformer(
    transformers=[('encoder', OneHotEncoder(drop='first'), ['State'])],
    remainder='passthrough'
)

X = ct.fit_transform(X)

model = LinearRegression()
model.fit(X, y)

new_company = pd.DataFrame({
    'R&D Spend': [91694.48],
    'Administration': [515841.3],
    'Marketing Spend': [11931.24],
    'State': ['Florida']
})

new_company_transformed = ct.transform(new_company)

predicted_profit = model.predict(new_company_transformed)

print(predicted_profit[0])
```
